```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.mixture import GaussianMixture
```

```python
np.random.seed(42)
segment1 = np.random.normal(loc=30, scale=5,size=100)
segment2 = np.random.normal(loc=60, scale=10,size=100)
segment3 = np.random.normal(loc=90, scale=8,size=100)
data= np.concatenate([segment1, segment2, segment3]).reshape(-1, 1)
```

```python
scaler= StandardScaler()
data_scaled= scaler.fit_transform(data)
n_components = 3
gmm = GaussianMixture(n_components = n_components, random_state=42)
gmm.fit(data_scaled)
```

```python
cluster_labels= gmm.predict(data_scaled)
```

```python
plt.scatter(data, np.zeros_like(data), c=cluster_labels, cmap='viridis')
plt.title('Cluster Segmentation')
plt.xlabel('Purchase Amount')
plt.show()
```

```python
# user testing
user_input = float(input("Enter a purchase amount to predict the customer segment: "))
user_input_scaled= scaler.fit_transform(np.array([[user_input]]))
```

```python
predicted_label = gmm.predict(user_input_scaled.reshape(-1, 1))[0]
print(f"the predicted customer egment for a purchase amount of {user_input} is {predicted_label + 1}")
```
