```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
data= {'Age': [20,28,27,38,36, 39] ,
       'MonthlyCharge': [50, 60, 80,54,100, 120],
       'churn': [ 1,0,1,0,0,1] }
df = pd.DataFrame(data)
df.head()
```

```python
X = df[['Age', 'MonthlyCharge']]
Y= df['churn']
```

```python
#s step 3
X_train , X_test , Y_train, Y_test = train_test_split(X , Y, test_size = 0.2, random_state= 42)
```

```python
# step 4
svc_model = SVC(kernel ='linear', C=1.0)
svc_model.fit(X_train, Y_train)
```

```python
# step 5
y_pred  = svc_model.predict(X_test)
```

```python

accuracy = accuracy_score(Y_test, y_pred)
```

```python
report = classification_report(Y_test, y_pred)
```

```python
user_age = float(input("Enter age: "))
user_monthly_charge = float(input("Enter monthly charges: "))
```

```python
user_input = np.array([[user_age, user_monthly_charge ]])
prediction = svc_model.predict(user_input)

if (prediction == 0):
    print("customer is likely to stay ")
else:
    print("customer is likely to churn ")
```
