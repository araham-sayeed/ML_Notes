```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split # for train-test model
from sklearn.linear_model import LinearRegression # for linear regression
```

```python
data = {'Hour_Study': [2,3,2,4,3,2,5,4,6,4,3,9] , 'Exam_Score': [50, 79, 73, 73, 54, 67, 90, 81, 58, 91, 78, 89] }
```

```python
df = pd.DataFrame(data)
```

```python
df.to_csv('student.csv')
```

```python
df
```

```python
X= df[['Hour_Study']]
Y= df['Exam_Score']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python

model =LinearRegression()
model.fit(X_train, Y_train)
```

```python
user_input= float(input("Enter the number of hour you studied: "))
```

```python
predicted_score= model.predict([[user_input]])
```

```python
print(f'Predicted Score = {predicted_score[0]: .2f}')
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

X = np.array([[25,30,0,] , [30, 40, 1] , [29, 38, 1], [25, 36, 0]])
Y = np.array([0 ,1, 0, 1])
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
accuracy= model.score(X_test, Y_test)
accuracy
```

```python
user_age= int(input("enter age: "))
user_time_spent=float(input("enter time spent on website: "))
user_added_to_cart = int(input("Enter 1 if added to cart: "))
user_data= np.array([[user_age, user_time_spent, user_added_to_cart]])
prediction = model.predict(user_data)
if prediction [0] == 1:
      print("customer is likely to purchase: ")
else:
      print("customer is not likely to purchase: ")
```
