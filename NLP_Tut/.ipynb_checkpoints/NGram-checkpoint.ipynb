```python
corpus = ["thor ate pizza",
          "Loki is tall",
          "Loki is eating pizza"]
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download necessary NLTK resources
nltk.download('all')
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words("english"))
```

```python
def preprocess(text):
    tokens = word_tokenize(text.lower())  # Tokenization and lowercasing
    filtered_tokens = [lemmatizer.lemmatize(word, pos="v") for word in tokens if word not in stop_words]  # Lemmatization & stopwords removal
    return " ".join(filtered_tokens)
```

```python
preprocessed_corpus = [preprocess(sentence) for sentence in corpus]
print(preprocessed_corpus)
```

```python
v = CountVectorizer(ngram_range=(1, 2))
```

```python
v.fit(preprocessed_corpus)
```

```python
v.vocabulary_
```

```python
v.transform(["Thor eat pizza"]).toarray()
```

```python
v.transform(["Hulk eat pizza"]).toarray()
```

```python
import pandas as pd

df = pd.read_json("newz_dataset.json")
df.head()
```

```python
df.category.value_counts()
```

```python
min_samples= 1
df_business = df[df.category == "Health"].sample(min_samples, random_state=1)
```

```python
df_business
```

```python
df_sports = df[df.category=="Sports"].sample(min_samples, random_state=2022)
df_crime = df[df.category=="Technology"].sample(min_samples, random_state=2022)
df_science = df[df.category=="Politics"].sample(min_samples, random_state=2022)
```

```python
df_balanced = pd.concat([df_business,df_sports,df_crime,df_science],axis=0)
df_balanced.category.value_counts()
```

```python
target = {'Health': 0, 'Sports': 1, 'Technology': 2, 'Politics': 3}

df_balanced['category_num'] = df_balanced['category'].map({
    'Health': 0,
    'Sports': 1, 
    'Technology': 2, 
    'Politics': 3
})
```
