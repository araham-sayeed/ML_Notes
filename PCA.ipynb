# Principle Component Analysis 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.decomposition import PCA
```

```python
data = {'Maths': [87, 92, 77, 73, 89, 34, 98, 41],
        'Science':[81, 77, 67, 90, 67, 61, 58, 81],
        'English': [91, 78, 57, 78, 56, 59, 89, 91] }
```

```python
# step1: creating a dataframe
df = pd.DataFrame(data)
```

```python
df
```

```python
# step 2: performing pca and fit
pca= PCA(n_components=1)
principle= pca.fit_transform(df)
```

```python
# step 3: principle component to data frame
df['principle']= principle
```

```python
df
```

```python
# step 4: visualization
math_score= df['Maths']
science_score= df['Science']
english_score= df['English']
principle= df['principle']
fig= plt.figure(figsize=(12,6))
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(math_score, science_score, english_score)
ax1.set_xlabel('Math Score')
ax1.set_ylabel('Science Score')
ax1.set_zlabel('English Score')
ax1.set_title('Original Score')

ax2 =fig.add_subplot(122)
ax2.scatter(principle, np.zeros_like(principle))
ax2.set_xlabel('principle')
ax2.set_title('First Principal Component')
plt.tight_layout()
plt.show()
```
