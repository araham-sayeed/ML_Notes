# Classification

##### Naive Bayes:  A probabilistic classifier based on Bayes’ Theorem assuming feature independence.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)

model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="Blues")
plt.title("Naive Bayes - Confusion Matrix")
plt.show()
```

##### Softmax: A generalization of logistic regression for multiclass problems.

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=200)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="Purples")
plt.title("Softmax Regression - Confusion Matrix")
plt.show()
```

##### K-nearest neighbors: Classifies a data point based on the majority class of its k nearest neighbors.

```python
#example
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="Greens")
plt.title("KNN (k=3) - Confusion Matrix")
plt.show()
```

##### Logistic Regression: A linear model used for binary classification based on sigmoid function.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Using only 2 classes and 2 features
X_bin = iris.data[iris.target != 2][:, :2]
y_bin = iris.target[iris.target != 2]

X_train, X_test, y_train, y_test = train_test_split(X_bin, y_bin, test_size=0.2, random_state=42)

model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="Reds")
plt.title("Binary Logistic Regression - Confusion Matrix")
plt.show()
```

##### Neural Network: A multi-layer feedforward network that learns complex patterns.

```python
from sklearn.neural_network import MLPClassifier

model = MLPClassifier(hidden_layer_sizes=(10,), max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="coolwarm")
plt.title("Neural Network - Confusion Matrix")
plt.show()
```
