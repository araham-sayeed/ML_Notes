```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from apyori import apriori
```

```python
store_data = pd.read_csv(r"C:\Users\Sanju\OneDrive\Desktop\3rd year\DE\store_data.csv", header=None)
```

```python
display(store_data.head())
```

```python
print(store_data.shape)
```

```python
records = []
for i in range(1, 7501):
    records.append([str(store_data.values[i, j]) for j in range(0, 20)])
```

```python
records
```

```python
type(records)
```

```python
association_rules = apriori(records, min_support=0.0045, min_confidence=0.2, min_lift=3, min_length=2)
```

```python
association_results = list(association_rules)
```

```python
for item in association_results:
    pair = item[0]
    items = [x for x in pair]
    print("Rule: " + items[0] + " -> " + items[1])
```

```python
# Additional
for item in association_results:
    print("Support: " + str(item[1]))
    print("Confidence: " + str(item[2][0][2]))
    print("Lift: " + str(item[2][0][3]))
    print("=====================================")
```

```python
for result in association_results:
    print("Items:", result.items)
    print("Support:", result.support)
    
    for ordered_stat in result.ordered_statistics:
        print("Base Item:", ordered_stat.items_base)
        print("Add Item:", ordered_stat.items_add)
        print("Confidence:", ordered_stat.confidence)
        print("Lift:", ordered_stat.lift)
```

```python
# Visualization in graph
supports = [result.support for result in association_results]

plt.bar(range(len(supports)), supports)
plt.xlabel('Rule Index')
plt.ylabel('Support')
plt.title('Support of Association Rules')
plt.show()
```

```python
results_list = []
for result in association_results:
    for ordered_stat in result.ordered_statistics:
        results_list.append({
            'Base': list(ordered_stat.items_base),
            'Add': list(ordered_stat.items_add),
            'Support': result.support,
            'Confidence': ordered_stat.confidence,
            'Lift': ordered_stat.lift,
        })
```

```python
df_result= pd.DataFrame(results_list)
```
